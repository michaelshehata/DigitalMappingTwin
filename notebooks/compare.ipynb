```python
import rasterio
import matplotlib.pyplot as plt

with rasterio.open("data/norwich_cover/norwich_2021.tif") as src:
    real = src.read(1)

with rasterio.open("outputs/future_2100.tif") as src:
    future = src.read(1)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.title("2021")
plt.imshow(real)

plt.subplot(1,2,2)
plt.title("2100 (Simulated)")
plt.imshow(future)

plt.show()
```
